```python
GSET_PARTITIONS = {
    1: ["G1", "G2", "G3", "G4", "G5", "G6", "G7", "G8", "G9", "G10"],
    2: ["G11", "G12", "G13"],
    3: ["G14", "G15", "G16", "G17", "G18", "G19", "G20", "G21"],
    4: ["G22", "G23", "G24", "G25", "G26", "G27", "G28", "G29", "G30", "G31"],
    5: ["G32", "G33", "G34"],
    6: ["G35", "G36", "G37", "G38", "G39", "G40", "G41", "G42"],
    7: ["G43", "G44", "G45", "G46", "G47"],
    8: ["G48", "G49", "G50"],
    9: ["G51", "G52", "G53", "G54"],
    10: ["G55", "G56"],
    11: ["G57"],
    12: ["G58", "G59"],
    13: ["G60", "G61"],
    14: ["G62"],
    15: ["G63", "G64"],
    16: ["G65"],
    17: ["G66"],
    18: ["G67", "G72"],
    19: ["G70"],
    20: ["G77"],
    21: ["G81"],
}

GSET_GRAPH_TYPES_BY_PARTITION = {
    1: "random",
    2: "toroidal",
    3: "plain",
    4: "random",
    5: "toroidal",
    6: "plain",
    7: "random",
    8: "toroidal",
    9: "plain",
    10: "plain",
    11: "toroid",
    12: "random",
    13: "plain",
    14: "toroid",
    15: "random",
    16: "toroid",
    17: "toroid",
    18: "toroid",
    19: "plain",
    20: "toroid",
    21: "toroid",
}


# ------------------------------------------------------------
# Storage mode by partition
# ------------------------------------------------------------

GSET_STORAGE_MODE_BY_PARTITION = {}

for partition_id in GSET_PARTITIONS:
    if partition_id <= 9:
        GSET_STORAGE_MODE_BY_PARTITION[partition_id] = "dense"
    else:
        GSET_STORAGE_MODE_BY_PARTITION[partition_id] = "edge_list"



# partition = batch number (1-21)
# graph = gset #
# graph type = plain/ toroid/ random
# storage mode = dense/ edge list
```

```python
# ------------------------------------------------------------
# Build graph-level target table
# ------------------------------------------------------------

target_rows = []

for partition_id, graphs in GSET_PARTITIONS.items():
    graph_type = GSET_GRAPH_TYPES_BY_PARTITION[partition_id]
    storage_mode = GSET_STORAGE_MODE_BY_PARTITION[partition_id]

    for graph in graphs:
        graph_path = GSET_DIR / graph

        target_rows.append(
            {
                "partition": partition_id,
                "graph": graph,
                "graph_type": graph_type,
                "storage_mode": storage_mode,
                "path": graph_path,
                "exists": graph_path.exists(),
            }
        )

targets_df = pd.DataFrame(target_rows)

# ------------------------------------------------------------
# Print partition-level overview
# ------------------------------------------------------------

print("\nPartition overview:")

for partition_id, graphs in GSET_PARTITIONS.items():
    graph_type = GSET_GRAPH_TYPES_BY_PARTITION[partition_id]
    storage_mode = GSET_STORAGE_MODE_BY_PARTITION[partition_id]

    print(
        f"Partition {partition_id:2d}: "
        f"{graphs[0]}-{graphs[-1]} | "
        f"{len(graphs):2d} instances | "
        f"type={graph_type:} | "
        #f"path ={graph_path} | "
        f"storage={storage_mode}"
    )
```

```python
# ------------------------------------------------------------
# Print graph-level table
# ------------------------------------------------------------

print("\nGraph-level target table:")
display(targets_df)


# ------------------------------------------------------------
# File checks
# ------------------------------------------------------------

missing_graph_files = targets_df[
    targets_df["exists"] == False
]["graph"].tolist()

print("\nMissing graph files:")
print(missing_graph_files)

assert len(missing_graph_files) == 0, (
    f"Missing graph files: {missing_graph_files}"
)


# ------------------------------------------------------------
# Basic consistency checks
# ------------------------------------------------------------

all_graphs = targets_df["graph"].tolist()
unique_graphs = sorted(set(all_graphs))

EXPECTED_NUM_GSET_GRAPHS = sum(
    len(graphs)
    for graphs in GSET_PARTITIONS.values()
)

assert len(all_graphs) == EXPECTED_NUM_GSET_GRAPHS, (
    f"Expected {EXPECTED_NUM_GSET_GRAPHS} graph entries, got {len(all_graphs)}."
)

assert len(unique_graphs) == EXPECTED_NUM_GSET_GRAPHS, (
    "Duplicate graph names detected in GSET_PARTITIONS."
)

assert set(GSET_PARTITIONS.keys()) == set(range(1, 22)), (
    "Partitions should be exactly 1 through 21."
)

assert set(GSET_GRAPH_TYPES_BY_PARTITION.keys()) == set(GSET_PARTITIONS.keys()), (
    "Graph-type partition keys do not match GSET_PARTITIONS."
)

assert set(GSET_STORAGE_MODE_BY_PARTITION.keys()) == set(GSET_PARTITIONS.keys()), (
    "Storage-mode partition keys do not match GSET_PARTITIONS."
)

assert (
    targets_df[targets_df["partition"] <= 9]["storage_mode"] == "dense"
).all(), (
    "Partitions 1-9 should use dense storage."
)

assert (
    targets_df[targets_df["partition"] >= 10]["storage_mode"] == "edge_list"
).all(), (
    "Partitions 10-21 should use edge-list storage."
)
```

```python
assert "targets_df" in globals(), (
    "Run Cell 01 first: targets_df is missing."
)

assert "GSET_DIR" in globals(), (
    "Run Cell 01 first: GSET_DIR is missing."
)


# ------------------------------------------------------------
# Locate validation file
# ------------------------------------------------------------

GSET_BEST_KNOWN_PATH = Path("gset_validation.txt")

if not GSET_BEST_KNOWN_PATH.exists():
    GSET_BEST_KNOWN_PATH = Path("../gset_validation.txt")

if not GSET_BEST_KNOWN_PATH.exists():
    GSET_BEST_KNOWN_PATH = GSET_DIR / "gset_validation.txt"

if not GSET_BEST_KNOWN_PATH.exists():
    GSET_BEST_KNOWN_PATH = Path("../data-gset/gset_validation.txt")


print("=" * 80)
print("BEST-KNOWN GSET VALIDATION VALUES")
print("=" * 80)

print("Validation path:")
print(GSET_BEST_KNOWN_PATH)

print("\nFile exists?")
print(GSET_BEST_KNOWN_PATH.exists())

assert GSET_BEST_KNOWN_PATH.exists(), (
    "Could not find gset_validation.txt. "
    "Set GSET_BEST_KNOWN_PATH manually."
)


# ------------------------------------------------------------
# Read validation file
# ------------------------------------------------------------

best_known_rows = []

with GSET_BEST_KNOWN_PATH.open("r") as f:
    for line_no, line in enumerate(f, start=1):
        line = line.strip()

        if not line:
            continue

        if line.startswith("#"):
            continue

        parts = line.replace(",", " ").split()

        if len(parts) < 2:
            continue

        graph_raw = parts[0]
        cut_raw = parts[1]

        if graph_raw.lower() in ["graph", "instance", "name"]:
            continue

        if graph_raw.startswith("G"):
            graph = graph_raw
        else:
            graph = "G" + graph_raw

        best_known_cut = float(cut_raw)

        best_known_rows.append(
            {
                "graph": graph,
                "best_known_cut": best_known_cut,
            }
        )

best_known_df = pd.DataFrame(best_known_rows)


# ------------------------------------------------------------
# Attach validation values to target table
# ------------------------------------------------------------

targets_with_validation_df = targets_df.merge(
    best_known_df,
    on="graph",
    how="left",
)
```

```python
from pathlib import Path
import pandas as pd

# ------------------------------------------------------------
# Input files
# ------------------------------------------------------------

V1_SMALL_RESULTS_PATH = Path("qefem_anneal_G1_G54_final_results.csv")
V1_BIG_RESULTS_PATH = Path("qefem_g55-g81_summary.csv")

print("=" * 80)
print("BUILDING CLEAN V1 BENCHMARK TABLE")
print("=" * 80)

print("Small V1 file:")
print(V1_SMALL_RESULTS_PATH.resolve())
print("exists:", V1_SMALL_RESULTS_PATH.exists())

print("\nBig V1 file:")
print(V1_BIG_RESULTS_PATH.resolve())
print("exists:", V1_BIG_RESULTS_PATH.exists())


# ------------------------------------------------------------
# Load small V1 results: G1-G54
# ------------------------------------------------------------

v1_small_raw_df = pd.read_csv(V1_SMALL_RESULTS_PATH)

v1_small_benchmark_df = v1_small_raw_df[
    [
        "instance",
        "nodes",
        "edges",
        "n_steps",
        "replicas",
        "lr",
        "best_cut",
        "best_value",
        "gap",
        "percent_accuracy",
        "best_replica",
        "runtime_sec",
    ]
].copy()

v1_small_benchmark_df = v1_small_benchmark_df.rename(
    columns={
        "best_cut": "v1_best",
        "best_value": "historical_best",
        "gap": "v1_gap",
        "percent_accuracy": "v1_percent_accuracy",
        "best_replica": "v1_best_replica",
        "runtime_sec": "v1_runtime_sec",
    }
)


# ------------------------------------------------------------
# Load big V1 results: G55-G81 selected subset
# ------------------------------------------------------------

v1_big_raw_df = pd.read_csv(V1_BIG_RESULTS_PATH)

v1_big_graph_df = v1_big_raw_df[
    v1_big_raw_df["graph"].notna()
].copy()

v1_big_benchmark_df = pd.DataFrame(
    {
        "instance": v1_big_graph_df["graph"],
        "nodes": v1_big_graph_df["N"].astype(int),
        "edges": v1_big_graph_df["M"].astype(int),
        "n_steps": v1_big_graph_df["steps"].astype(int),
        "replicas": v1_big_graph_df["replicas"].astype(int),
        "lr": v1_big_graph_df["lr"],
        "v1_best": v1_big_graph_df["qefem_best_cut"],
        "historical_best": v1_big_graph_df["best_known_cut"],
        "v1_gap": v1_big_graph_df["gap"],
        "v1_percent_accuracy": v1_big_graph_df["accuracy_percent"],
        "v1_best_replica": v1_big_graph_df["best_replica"].astype(int),
        "v1_runtime_sec": v1_big_graph_df["runtime_sec"],
    }
)


# ------------------------------------------------------------
# Combine small and big V1 benchmark tables
# ------------------------------------------------------------

benchmark_df = pd.concat(
    [
        v1_small_benchmark_df,
        v1_big_benchmark_df,
    ],
    ignore_index=True,
)

benchmark_df["graph_number"] = (
    benchmark_df["instance"]
    .str.replace("G", "", regex=False)
    .astype(int)
)

benchmark_df = benchmark_df.sort_values(
    "graph_number"
).reset_index(drop=True)


# ------------------------------------------------------------
# Add graph type and storage mode from targets_df
# ------------------------------------------------------------

benchmark_df = benchmark_df.merge(
    targets_df[
        [
            "graph",
            "partition",
            "graph_type",
            "storage_mode",
        ]
    ].rename(
        columns={
            "graph": "instance",
        }
    ),
    on="instance",
    how="left",
)

benchmark_df = benchmark_df[
    [
        "partition",
        "instance",
        "graph_type",
        "storage_mode",
        "nodes",
        "edges",
        "n_steps",
        "replicas",
        "lr",
        "v1_best",
        "historical_best",
        "v1_gap",
        "v1_percent_accuracy",
        "v1_best_replica",
        "v1_runtime_sec",
    ]
].copy()


# ------------------------------------------------------------
# Recompute simple consistency columns internally, but do not keep them
# ------------------------------------------------------------

gap_check = benchmark_df["historical_best"] - benchmark_df["v1_best"]
accuracy_check = 100.0 * benchmark_df["v1_best"] / benchmark_df["historical_best"]

max_gap_error = (gap_check - benchmark_df["v1_gap"]).abs().max()
max_accuracy_error = (accuracy_check - benchmark_df["v1_percent_accuracy"]).abs().max()

print("\nConsistency:")
print("max gap error:", float(max_gap_error))
print("max accuracy error:", float(max_accuracy_error))
```

```python
# ------------------------------------------------------------
# Display clean table
# ------------------------------------------------------------

print("\nClean benchmark table:")
display(benchmark_df)


# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

BENCHMARK_PATH = Path("qefem_v1_benchmark_table.csv")

benchmark_df.to_csv(
    BENCHMARK_PATH,
    index=False,
)

print("\nSaved benchmark table:")
print(BENCHMARK_PATH.resolve())
```

```python
import math
import time
from pathlib import Path

import torch
import pandas as pd


if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

DTYPE = torch.float32


N_REPLICAS = 128
N_STEPS = 2000
LOG_EVERY = 50
SEED = 7

OPTIMIZER_NAME = "Adam"
ADAM_LR = 0.03

T_MAX = 1.16
T_MIN = 6e-5


GAMMA_MAX = 1.0
GAMMA_MIN = 0.0

GAMMA_PULSE_AMP = 0.05
GAMMA_PULSE_CENTER_FRAC = 0.80
GAMMA_PULSE_WIDTH_FRAC = 0.03


R_INIT = 1e-2
RAW_R_NOISE = 1e-3

THETA_CENTER = math.pi / 2
THETA_NOISE = 1e-2


QEFEM_V2_OUTPUT_DIR = Path("qefem_v2_outputs")
QEFEM_V2_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ------------------------------------------------------------
# Print settings
# ------------------------------------------------------------

print("=" * 80)
print("QEFEM V2 SETTINGS")
print("=" * 80)

print("device:", device)
print("dtype :", DTYPE)

print("\nOptimizer:")
print("optimizer:", OPTIMIZER_NAME)
print("lr:", ADAM_LR)

print("\nRun size:")
print("replicas:", N_REPLICAS)
print("steps:", N_STEPS)
print("log every:", LOG_EVERY)
print("seed:", SEED)

print("\nTemperature:")
print("T_MAX:", T_MAX)
print("T_MIN:", T_MIN)

print("\nGamma:")
print("GAMMA_MAX:", GAMMA_MAX)
print("GAMMA_MIN:", GAMMA_MIN)
print("GAMMA_PULSE_AMP:", GAMMA_PULSE_AMP)
print("GAMMA_PULSE_CENTER_FRAC:", GAMMA_PULSE_CENTER_FRAC)
print("GAMMA_PULSE_WIDTH_FRAC:", GAMMA_PULSE_WIDTH_FRAC)

print("\nInitialization:")
print("R_INIT:", R_INIT)
print("RAW_R_NOISE:", RAW_R_NOISE)
print("THETA_CENTER:", THETA_CENTER)
print("THETA_NOISE:", THETA_NOISE)

print("\nOutput directory:")
print(QEFEM_V2_OUTPUT_DIR.resolve())


# ------------------------------------------------------------
# Minimal sanity checks
# ------------------------------------------------------------

assert N_REPLICAS > 0
assert N_STEPS > 0
assert LOG_EVERY > 0
assert ADAM_LR > 0

assert T_MAX > T_MIN > 0
assert GAMMA_MAX >= GAMMA_MIN >= 0

assert GAMMA_PULSE_AMP >= 0
assert 0.0 <= GAMMA_PULSE_CENTER_FRAC <= 1.0
assert GAMMA_PULSE_WIDTH_FRAC > 0

assert 0 < R_INIT < 1
assert RAW_R_NOISE >= 0
assert THETA_NOISE >= 0

assert "benchmark_df" in globals()
assert len(benchmark_df) > 0
```

```python
step_index = torch.arange(
    N_STEPS,
    device=device,
    dtype=DTYPE,
)


# ------------------------------------------------------------
# Temperature and beta
# ------------------------------------------------------------

T_schedule = torch.linspace(
    T_MAX,
    T_MIN,
    N_STEPS,
    device=device,
    dtype=DTYPE,
)

beta_schedule = 1.0 / T_schedule


# ------------------------------------------------------------
# Gamma = linear anneal + Gaussian pulse
# ------------------------------------------------------------

gamma_base_schedule = torch.linspace(
    GAMMA_MAX,
    GAMMA_MIN,
    N_STEPS,
    device=device,
    dtype=DTYPE,
)

gamma_pulse_center = GAMMA_PULSE_CENTER_FRAC * (N_STEPS - 1)
gamma_pulse_width = GAMMA_PULSE_WIDTH_FRAC * N_STEPS

gamma_pulse_schedule = GAMMA_PULSE_AMP * torch.exp(
    -0.5 * (
        (step_index - gamma_pulse_center) / gamma_pulse_width
    ) ** 2
)

gamma_schedule = gamma_base_schedule + gamma_pulse_schedule

# Force final transverse field to zero for final hard readout.
gamma_schedule[-1] = GAMMA_MIN


# ------------------------------------------------------------
# Schedule dataframe
# ------------------------------------------------------------

schedule_df = pd.DataFrame(
    {
        "step": torch.arange(1, N_STEPS + 1).detach().cpu().numpy(),
        "temperature": T_schedule.detach().cpu().numpy(),
        "beta": beta_schedule.detach().cpu().numpy(),
        "gamma_base": gamma_base_schedule.detach().cpu().numpy(),
        "gamma_pulse": gamma_pulse_schedule.detach().cpu().numpy(),
        "gamma": gamma_schedule.detach().cpu().numpy(),
    }
)


# ------------------------------------------------------------
# Print compact summary
# ------------------------------------------------------------

print("=" * 80)
print("ANNEALING SCHEDULES")
print("=" * 80)

print("steps:", N_STEPS)

print("\nTemperature:")
print("first :", float(T_schedule[0].detach().cpu()))
print("middle:", float(T_schedule[N_STEPS // 2].detach().cpu()))
print("last  :", float(T_schedule[-1].detach().cpu()))

print("\nBeta:")
print("first :", float(beta_schedule[0].detach().cpu()))
print("middle:", float(beta_schedule[N_STEPS // 2].detach().cpu()))
print("last  :", float(beta_schedule[-1].detach().cpu()))

print("\nGamma:")
print("first :", float(gamma_schedule[0].detach().cpu()))
print("middle:", float(gamma_schedule[N_STEPS // 2].detach().cpu()))
print("max   :", float(gamma_schedule.max().detach().cpu()))
print("last  :", float(gamma_schedule[-1].detach().cpu()))

print("\nGaussian pulse:")
print("center step:", float(gamma_pulse_center))
print("width steps:", float(gamma_pulse_width))
print("max pulse  :", float(gamma_pulse_schedule.max().detach().cpu()))
```

```python
# ------------------------------------------------------------
# Bloch parametrization
# ------------------------------------------------------------

def bloch_from_raw(raw_R, theta):
    R = torch.sigmoid(raw_R)

    rx = R * torch.sin(theta)
    rz = R * torch.cos(theta)

    return R, rx, rz


# ------------------------------------------------------------
# Quantum entropy from Bloch length
# ------------------------------------------------------------

def quantum_entropy_from_R(R, eps=1e-12):
    lambda_plus = 0.5 * (1.0 + R)
    lambda_minus = 0.5 * (1.0 - R)

    lambda_plus = torch.clamp(
        lambda_plus,
        min=eps,
        max=1.0,
    )

    lambda_minus = torch.clamp(
        lambda_minus,
        min=eps,
        max=1.0,
    )

    entropy_per_site = -(
        lambda_plus * torch.log(lambda_plus)
        +
        lambda_minus * torch.log(lambda_minus)
    )

    entropy = entropy_per_site.sum(dim=1)

    return entropy


# ------------------------------------------------------------
# Transverse driver energy
# ------------------------------------------------------------

def transverse_driver_energy(rx, gamma):
    gamma = torch.as_tensor(
        gamma,
        device=rx.device,
        dtype=rx.dtype,
    )

    driver_energy = -gamma * rx.sum(dim=1)

    return driver_energy


# ------------------------------------------------------------
# Dense symmetric-W MaxCut functions
# ------------------------------------------------------------

def dense_expected_cut(W, rz):
    quad = torch.einsum(
        "ri,ij,rj->r",
        rz,
        W,
        rz,
    )

    expected_cut = 0.25 * (
        W.sum() - quad
    )

    return expected_cut


def dense_hard_cut(W, spins):
    single_input = False

    if spins.dim() == 1:
        spins = spins.unsqueeze(0)
        single_input = True

    spins = spins.to(
        device=W.device,
        dtype=W.dtype,
    )

    quad = torch.einsum(
        "ri,ij,rj->r",
        spins,
        W,
        spins,
    )

    cut = 0.25 * (
        W.sum() - quad
    )

    if single_input:
        cut = cut.squeeze(0)

    return cut


def dense_qefem_free_energy(W, raw_R, theta, beta, gamma):
    beta = torch.as_tensor(
        beta,
        device=raw_R.device,
        dtype=raw_R.dtype,
    )

    R, rx, rz = bloch_from_raw(
        raw_R,
        theta,
    )

    expected_cut = dense_expected_cut(
        W=W,
        rz=rz,
    )

    entropy = quantum_entropy_from_R(R)

    driver_energy = transverse_driver_energy(
        rx=rx,
        gamma=gamma,
    )

    free_energy = (
        -expected_cut
        + driver_energy
        - entropy / beta
    )

    aux = {
        "R": R,
        "rx": rx,
        "rz": rz,
        "expected_cut": expected_cut,
        "entropy": entropy,
        "driver_energy": driver_energy,
        "free_energy": free_energy,
    }

    return free_energy, aux


# ------------------------------------------------------------
# Edge-list MaxCut functions
# ------------------------------------------------------------

def edge_expected_cut(edge_i, edge_j, edge_w, rz):
    zi = rz[:, edge_i]
    zj = rz[:, edge_j]

    expected_cut = 0.5 * (
        edge_w * (1.0 - zi * zj)
    ).sum(dim=1)

    return expected_cut


def edge_hard_cut(edge_i, edge_j, edge_w, spins):
    single_input = False

    if spins.dim() == 1:
        spins = spins.unsqueeze(0)
        single_input = True

    spins = spins.to(
        device=edge_w.device,
        dtype=edge_w.dtype,
    )

    si = spins[:, edge_i]
    sj = spins[:, edge_j]

    cut = 0.5 * (
        edge_w * (1.0 - si * sj)
    ).sum(dim=1)

    if single_input:
        cut = cut.squeeze(0)

    return cut


def edge_qefem_free_energy(edge_i, edge_j, edge_w, raw_R, theta, beta, gamma):
    beta = torch.as_tensor(
        beta,
        device=raw_R.device,
        dtype=raw_R.dtype,
    )

    R, rx, rz = bloch_from_raw(
        raw_R,
        theta,
    )

    expected_cut = edge_expected_cut(
        edge_i=edge_i,
        edge_j=edge_j,
        edge_w=edge_w,
        rz=rz,
    )

    entropy = quantum_entropy_from_R(R)

    driver_energy = transverse_driver_energy(
        rx=rx,
        gamma=gamma,
    )

    free_energy = (
        -expected_cut
        + driver_energy
        - entropy / beta
    )

    aux = {
        "R": R,
        "rx": rx,
        "rz": rz,
        "expected_cut": expected_cut,
        "entropy": entropy,
        "driver_energy": driver_energy,
        "free_energy": free_energy,
    }

    return free_energy, aux


# ------------------------------------------------------------
# Initialization
# ------------------------------------------------------------

def initialize_qefem_variables(
    N,
    replicas,
    R_init,
    raw_R_noise,
    theta_center,
    theta_noise,
    seed,
):
    torch.manual_seed(seed)

    R_init_tensor = torch.tensor(
        R_init,
        device=device,
        dtype=DTYPE,
    )

    raw_R_center = torch.log(
        R_init_tensor / (1.0 - R_init_tensor)
    )

    raw_R = raw_R_center + raw_R_noise * torch.randn(
        replicas,
        N,
        device=device,
        dtype=DTYPE,
    )

    theta = theta_center + theta_noise * torch.randn(
        replicas,
        N,
        device=device,
        dtype=DTYPE,
    )

    raw_R.requires_grad_(True)
    theta.requires_grad_(True)

    return raw_R, theta


# ------------------------------------------------------------
# Hard readout
# ------------------------------------------------------------

def hard_readout_from_rz(rz):
    spins = torch.where(
        rz >= 0,
        torch.ones_like(rz),
        -torch.ones_like(rz),
    )

    return spins


# ------------------------------------------------------------
# Storage-aware wrappers
# ------------------------------------------------------------

def qefem_free_energy_for_graph(graph_data, raw_R, theta, beta, gamma):
    storage_mode = graph_data["storage_mode"]

    if storage_mode == "dense":
        free_energy, aux = dense_qefem_free_energy(
            W=graph_data["W"],
            raw_R=raw_R,
            theta=theta,
            beta=beta,
            gamma=gamma,
        )

    elif storage_mode == "edge_list":
        free_energy, aux = edge_qefem_free_energy(
            edge_i=graph_data["edge_i"],
            edge_j=graph_data["edge_j"],
            edge_w=graph_data["edge_w"],
            raw_R=raw_R,
            theta=theta,
            beta=beta,
            gamma=gamma,
        )

    else:
        raise ValueError(
            f"Unknown storage_mode: {storage_mode}"
        )

    return free_energy, aux


def hard_cut_for_graph(graph_data, spins):
    storage_mode = graph_data["storage_mode"]

    if storage_mode == "dense":
        cut = dense_hard_cut(
            W=graph_data["W"],
            spins=spins,
        )

    elif storage_mode == "edge_list":
        cut = edge_hard_cut(
            edge_i=graph_data["edge_i"],
            edge_j=graph_data["edge_j"],
            edge_w=graph_data["edge_w"],
            spins=spins,
        )

    else:
        raise ValueError(
            f"Unknown storage_mode: {storage_mode}"
        )

    return cut


print("=" * 80)
print("QEFEM CORE FUNCTIONS READY")
print("=" * 80)

print("Functions:")
print("bloch_from_raw(raw_R, theta)")
print("quantum_entropy_from_R(R)")
print("dense_qefem_free_energy(W, raw_R, theta, beta, gamma)")
print("edge_qefem_free_energy(edge_i, edge_j, edge_w, raw_R, theta, beta, gamma)")
print("initialize_qefem_variables(...)")
print("hard_readout_from_rz(rz)")
print("qefem_free_energy_for_graph(graph_data, raw_R, theta, beta, gamma)")
print("hard_cut_for_graph(graph_data, spins)")
```

```python
# ------------------------------------------------------------
# Edge-list loader
# ------------------------------------------------------------

def load_gset_edge_list(instance):
    path = Path(GSET_DIR) / instance

    edge_i_list = []
    edge_j_list = []
    edge_w_list = []

    with path.open("r") as f:
        header = f.readline().strip().split()

        N = int(header[0])
        M = int(header[1])

        for line in f:
            line = line.strip()

            if not line:
                continue

            i, j, w = line.split()

            edge_i_list.append(int(i) - 1)
            edge_j_list.append(int(j) - 1)
            edge_w_list.append(float(w))

    edge_i = torch.tensor(
        edge_i_list,
        device=device,
        dtype=torch.long,
    )

    edge_j = torch.tensor(
        edge_j_list,
        device=device,
        dtype=torch.long,
    )

    edge_w = torch.tensor(
        edge_w_list,
        device=device,
        dtype=DTYPE,
    )

    graph_data = {
        "instance": instance,
        "storage_mode": "edge_list",
        "N": N,
        "M": M,
        "edge_i": edge_i,
        "edge_j": edge_j,
        "edge_w": edge_w,
    }

    return graph_data


# ------------------------------------------------------------
# Dense symmetric-W loader
# ------------------------------------------------------------

def load_gset_dense(instance):
    edge_graph_data = load_gset_edge_list(instance)

    N = edge_graph_data["N"]
    M = edge_graph_data["M"]

    edge_i = edge_graph_data["edge_i"]
    edge_j = edge_graph_data["edge_j"]
    edge_w = edge_graph_data["edge_w"]

    W = torch.zeros(
        N,
        N,
        device=device,
        dtype=DTYPE,
    )

    W[edge_i, edge_j] = edge_w
    W[edge_j, edge_i] = edge_w

    graph_data = {
        "instance": instance,
        "storage_mode": "dense",
        "N": N,
        "M": M,
        "W": W,
    }

    return graph_data


# ------------------------------------------------------------
# Storage-aware loader
# ------------------------------------------------------------

def load_graph_for_qefem(instance, storage_mode):
    if storage_mode == "dense":
        graph_data = load_gset_dense(instance)

    elif storage_mode == "edge_list":
        graph_data = load_gset_edge_list(instance)

    else:
        raise ValueError(
            f"Unknown storage_mode={storage_mode}"
        )

    return graph_data


# ------------------------------------------------------------
# Tiny loader preview
# ------------------------------------------------------------

preview_row = benchmark_df.iloc[0]

preview_graph_data = load_graph_for_qefem(
    instance=preview_row["instance"],
    storage_mode=preview_row["storage_mode"],
)

print("=" * 80)
print("GRAPH LOADERS READY")
print("=" * 80)

print("preview instance:", preview_graph_data["instance"])
print("storage mode    :", preview_graph_data["storage_mode"])
print("N               :", preview_graph_data["N"])
print("M               :", preview_graph_data["M"])

if preview_graph_data["storage_mode"] == "dense":
    print("W shape         :", tuple(preview_graph_data["W"].shape))
else:
    print("edge_i shape    :", tuple(preview_graph_data["edge_i"].shape))
    print("edge_j shape    :", tuple(preview_graph_data["edge_j"].shape))
    print("edge_w shape    :", tuple(preview_graph_data["edge_w"].shape))

del preview_graph_data

print("\nCell 08 complete: graph loaders are ready.")
```

```python
# ------------------------------------------------------------
# Main group-level runner
# ------------------------------------------------------------

def run_qefem_run_group(run_group_index):
    run_group_row = run_group_summary_df.iloc[run_group_index]

    run_group = run_group_row["run_group"]
    graphs = get_graphs_from_run_group_row(run_group_row)

    group_output_dir = QEFEM_V2_OUTPUT_DIR / run_group
    group_output_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    log_output_dir = group_output_dir / "logs"
    log_output_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    spin_output_dir = group_output_dir / "best_spins"
    spin_output_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    selected_benchmark_df = benchmark_df[
        benchmark_df["instance"].isin(graphs)
    ].copy()

    selected_benchmark_df["graph_order"] = (
        selected_benchmark_df["instance"]
        .str.replace("G", "", regex=False)
        .astype(int)
    )

    selected_benchmark_df = selected_benchmark_df.sort_values(
        "graph_order"
    ).reset_index(drop=True)

    selected_benchmark_df = selected_benchmark_df.drop(
        columns=["graph_order"]
    )

    print("=" * 80)
    print("RUNNING QEFEM V2 RUN GROUP")
    print("=" * 80)

    print("run_group_index:", run_group_index)
    print("run_group:", run_group)
    print("graphs:", graphs)
    print("num graphs:", len(graphs))

    print("\nOutput directory:")
    print(group_output_dir.resolve())

    print("\nSelected benchmark rows:")
    display(selected_benchmark_df)

    group_result_rows = []
    group_log_dfs = []

    for graph_idx, benchmark_row in selected_benchmark_df.iterrows():
        instance = benchmark_row["instance"]
        storage_mode = benchmark_row["storage_mode"]

        nodes = int(benchmark_row["nodes"])
        edges = int(benchmark_row["edges"])

        historical_best = float(benchmark_row["historical_best"])

        v1_best = float(benchmark_row["v1_best"])
        v1_gap = float(benchmark_row["v1_gap"])
        v1_percent_accuracy = float(benchmark_row["v1_percent_accuracy"])

        graph_seed = SEED + graph_idx

        print("\n" + "#" * 80)
        print(f"RUNNING {instance} | {storage_mode} | seed={graph_seed}")
        print("#" * 80)

        graph_start_time = time.time()

        graph_data = load_graph_for_qefem(
            instance=instance,
            storage_mode=storage_mode,
        )

        raw_R, theta = initialize_qefem_variables(
            N=nodes,
            replicas=N_REPLICAS,
            R_init=R_INIT,
            raw_R_noise=RAW_R_NOISE,
            theta_center=THETA_CENTER,
            theta_noise=THETA_NOISE,
            seed=graph_seed,
        )

        optimizer = torch.optim.Adam(
            [
                raw_R,
                theta,
            ],
            lr=ADAM_LR,
        )

        v2_best = -float("inf")
        v2_best_step = None
        v2_best_replica = None
        best_step_success_count = 0
        best_step_success_probability = 0.0
        best_spins = None

        ever_success_mask = torch.zeros(
            N_REPLICAS,
            device=device,
            dtype=torch.bool,
        )

        log_rows = []

        for step in range(N_STEPS):
            beta_t = beta_schedule[step]
            gamma_t = gamma_schedule[step]

            should_log = (
                step == 0
                or (step + 1) % LOG_EVERY == 0
                or step == N_STEPS - 1
            )

            optimizer.zero_grad()

            free_energy, aux = qefem_free_energy_for_graph(
                graph_data=graph_data,
                raw_R=raw_R,
                theta=theta,
                beta=beta_t,
                gamma=gamma_t,
            )

            loss = free_energy.mean()
            loss.backward()

            if should_log:
                with torch.no_grad():
                    raw_R_grad_norm = float(raw_R.grad.norm().detach().cpu())
                    theta_grad_norm = float(theta.grad.norm().detach().cpu())

                    raw_R_grad_mean_abs = float(raw_R.grad.abs().mean().detach().cpu())
                    theta_grad_mean_abs = float(theta.grad.abs().mean().detach().cpu())

                    raw_R_grad_max_abs = float(raw_R.grad.abs().max().detach().cpu())
                    theta_grad_max_abs = float(theta.grad.abs().max().detach().cpu())

                    raw_R_before = raw_R.detach().clone()
                    theta_before = theta.detach().clone()

            optimizer.step()

            with torch.no_grad():
                if should_log:
                    raw_R_step = raw_R.detach() - raw_R_before
                    theta_step = theta.detach() - theta_before

                    raw_R_step_norm = float(raw_R_step.norm().detach().cpu())
                    theta_step_norm = float(theta_step.norm().detach().cpu())

                    raw_R_step_mean_abs = float(raw_R_step.abs().mean().detach().cpu())
                    theta_step_mean_abs = float(theta_step.abs().mean().detach().cpu())

                eval_free_energy, eval_aux = qefem_free_energy_for_graph(
                    graph_data=graph_data,
                    raw_R=raw_R,
                    theta=theta,
                    beta=beta_t,
                    gamma=gamma_t,
                )

                R = eval_aux["R"]
                rx = eval_aux["rx"]
                rz = eval_aux["rz"]

                spins = hard_readout_from_rz(rz)

                cut_values = hard_cut_for_graph(
                    graph_data=graph_data,
                    spins=spins,
                )

                step_best_cut_tensor, step_best_replica_tensor = torch.max(
                    cut_values,
                    dim=0,
                )

                step_best_cut = float(step_best_cut_tensor.detach().cpu())
                step_best_replica = int(step_best_replica_tensor.detach().cpu())

                success_mask = cut_values >= historical_best

                step_success_count = int(
                    success_mask.sum().detach().cpu()
                )

                step_success_probability = step_success_count / N_REPLICAS

                ever_success_mask |= success_mask

                if step_best_cut > v2_best:
                    v2_best = step_best_cut
                    v2_best_step = step + 1
                    v2_best_replica = step_best_replica

                    best_step_success_count = step_success_count
                    best_step_success_probability = step_success_probability

                    best_spins = spins[
                        v2_best_replica
                    ].detach().clone()

                current_accuracy = 100.0 * v2_best / historical_best

                if should_log:
                    mean_R = float(R.mean().detach().cpu())
                    mean_abs_rx = float(rx.abs().mean().detach().cpu())
                    mean_abs_rz = float(rz.abs().mean().detach().cpu())

                    frac_polarized = float(
                        (rz.abs() > 0.95).to(DTYPE).mean().detach().cpu()
                    )

                    frac_ambiguous = float(
                        (rz.abs() < 0.05).to(DTYPE).mean().detach().cpu()
                    )

                    log_rows.append(
                        {
                            "run_group": run_group,
                            "instance": instance,
                            "step": step + 1,
                            "temperature": float(T_schedule[step].detach().cpu()),
                            "beta": float(beta_t.detach().cpu()),
                            "gamma": float(gamma_t.detach().cpu()),

                            "loss": float(loss.detach().cpu()),
                            "mean_free_energy": float(eval_aux["free_energy"].mean().detach().cpu()),
                            "mean_expected_cut": float(eval_aux["expected_cut"].mean().detach().cpu()),
                            "max_expected_cut": float(eval_aux["expected_cut"].max().detach().cpu()),
                            "mean_entropy": float(eval_aux["entropy"].mean().detach().cpu()),
                            "mean_driver_energy": float(eval_aux["driver_energy"].mean().detach().cpu()),

                            "step_best_cut": step_best_cut,
                            "global_best_cut": v2_best,
                            "global_best_accuracy": current_accuracy,

                            "step_success_count": step_success_count,
                            "step_success_probability": step_success_probability,
                            "ever_success_count": int(ever_success_mask.sum().detach().cpu()),
                            "ever_success_probability": float(ever_success_mask.sum().detach().cpu()) / N_REPLICAS,

                            "mean_R": mean_R,
                            "mean_abs_rx": mean_abs_rx,
                            "mean_abs_rz": mean_abs_rz,
                            "frac_polarized": frac_polarized,
                            "frac_ambiguous": frac_ambiguous,

                            "raw_R_grad_norm": raw_R_grad_norm,
                            "theta_grad_norm": theta_grad_norm,
                            "raw_R_grad_mean_abs": raw_R_grad_mean_abs,
                            "theta_grad_mean_abs": theta_grad_mean_abs,
                            "raw_R_grad_max_abs": raw_R_grad_max_abs,
                            "theta_grad_max_abs": theta_grad_max_abs,

                            "raw_R_step_norm": raw_R_step_norm,
                            "theta_step_norm": theta_step_norm,
                            "raw_R_step_mean_abs": raw_R_step_mean_abs,
                            "theta_step_mean_abs": theta_step_mean_abs,
                        }
                    )

                    print(
                        f"{instance} | "
                        f"step {step + 1:5d}/{N_STEPS} | "
                        f"best {v2_best:.1f} | "
                        f"acc {current_accuracy:.4f}% | "
                        f"success {step_success_count}/{N_REPLICAS} | "
                        f"gamma {float(gamma_t.detach().cpu()):.5f} | "
                        f"|rz| {mean_abs_rz:.4f}"
                    )

        graph_runtime_sec = time.time() - graph_start_time

        with torch.no_grad():
            final_free_energy, final_aux = qefem_free_energy_for_graph(
                graph_data=graph_data,
                raw_R=raw_R,
                theta=theta,
                beta=beta_schedule[-1],
                gamma=gamma_schedule[-1],
            )

            final_spins = hard_readout_from_rz(
                final_aux["rz"]
            )

            final_cut_values = hard_cut_for_graph(
                graph_data=graph_data,
                spins=final_spins,
            )

            final_best_cut = float(final_cut_values.max().detach().cpu())

            final_success_mask = final_cut_values >= historical_best

            final_success_count = int(
                final_success_mask.sum().detach().cpu()
            )

            final_success_probability = final_success_count / N_REPLICAS

        ever_success_count = int(
            ever_success_mask.sum().detach().cpu()
        )

        ever_success_probability = ever_success_count / N_REPLICAS

        v2_gap = historical_best - v2_best
        v2_percent_accuracy = 100.0 * v2_best / historical_best

        final_gap = historical_best - final_best_cut
        final_percent_accuracy = 100.0 * final_best_cut / historical_best

        result_row = {
            "run_group": run_group,
            "partition": int(benchmark_row["partition"]),
            "instance": instance,
            "graph_type": benchmark_row["graph_type"],
            "storage_mode": storage_mode,

            "nodes": nodes,
            "edges": edges,

            "n_steps": N_STEPS,
            "replicas": N_REPLICAS,
            "lr": ADAM_LR,
            "seed": graph_seed,

            "historical_best": historical_best,

            "v1_best": v1_best,
            "v1_gap": v1_gap,
            "v1_percent_accuracy": v1_percent_accuracy,

            "v2_best": v2_best,
            "v2_gap": v2_gap,
            "v2_percent_accuracy": v2_percent_accuracy,
            "v2_best_step": v2_best_step,
            "v2_best_replica": v2_best_replica,

            "final_best": final_best_cut,
            "final_gap": final_gap,
            "final_percent_accuracy": final_percent_accuracy,

            "best_step_success_count": best_step_success_count,
            "best_step_success_probability": best_step_success_probability,

            "ever_success_count": ever_success_count,
            "ever_success_probability": ever_success_probability,

            "final_success_count": final_success_count,
            "final_success_probability": final_success_probability,

            "v2_runtime_sec": graph_runtime_sec,
        }

        graph_log_df = pd.DataFrame(log_rows)

        group_result_rows.append(result_row)
        group_log_dfs.append(graph_log_df)

        graph_log_path = log_output_dir / f"{instance}_training_log.csv"
        graph_log_df.to_csv(
            graph_log_path,
            index=False,
        )

        best_spin_path = spin_output_dir / f"{instance}_best_spins.pt"
        torch.save(
            best_spins.detach().cpu(),
            best_spin_path,
        )

        print_gradient_mobility_summary(
            log_df=graph_log_df,
            instance=instance,
        )

        print_graph_result_summary(
            result_row=result_row,
        )

        partial_results_df = pd.DataFrame(group_result_rows)

        partial_results_path = group_output_dir / "partial_group_results.csv"
        partial_results_df.to_csv(
            partial_results_path,
            index=False,
        )

        print("\nSaved graph log:")
        print(graph_log_path.resolve())

        print("Saved best spins:")
        print(best_spin_path.resolve())

        print("Saved partial group results:")
        print(partial_results_path.resolve())

        del graph_data
        del raw_R
        del theta
        del optimizer

        if device.type == "cuda":
            torch.cuda.empty_cache()

    group_results_df = pd.DataFrame(group_result_rows)

    group_logs_df = pd.concat(
        group_log_dfs,
        ignore_index=True,
    )

    group_results_path = group_output_dir / "group_results.csv"
    group_logs_path = group_output_dir / "group_training_logs.csv"

    group_results_df.to_csv(
        group_results_path,
        index=False,
    )

    group_logs_df.to_csv(
        group_logs_path,
        index=False,
    )

    print("\n" + "=" * 80)
    print("RUN GROUP COMPLETE")
    print("=" * 80)

    print("run_group:", run_group)
    print("graphs:", len(group_results_df))

    print("\nGroup results:")
    display(group_results_df)

    compact_columns = [
        "instance",
        "historical_best",
        "v1_best",
        "v1_gap",
        "v1_percent_accuracy",
        "v2_best",
        "v2_gap",
        "v2_percent_accuracy",
        "v2_best_step",
        "v2_best_replica",
        "ever_success_count",
        "ever_success_probability",
        "final_best",
        "final_gap",
        "final_percent_accuracy",
        "final_success_count",
        "final_success_probability",
        "v2_runtime_sec",
    ]

    print("\nCompact group results:")
    display(group_results_df[compact_columns])

    print("\nSaved group results:")
    print(group_results_path.resolve())

    print("\nSaved group logs:")
    print(group_logs_path.resolve())

    return group_results_df, group_logs_df


print("=" * 80)
print("GROUP-LEVEL QEFEM RUNNER READY")
print("=" * 80)

print("Use:")
print("group_results_df, group_logs_df = run_qefem_run_group(run_group_index)")
```
